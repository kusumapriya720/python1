```python
#list in python 
#List is an ordered and changeable collection that can store multiple values
marks = [80, 90, 75, 85]

print(marks)
```

```python
#accessing the elements in a list 
marks = [80, 90, 75, 85]

print(marks[0])
print(marks[1])
print(marks[2])
```

```python
#changing elements in a list 
marks = [80, 90, 75]

marks[1] = 95

print(marks)
```

```python
#add elements to a list
marks = [80, 90, 75]

marks.append(85)

print(marks)
```

```python
#remove elements from a list
marks = [80, 90, 75]

marks.remove(90)

print(marks)
```

```python
a = [1, 2, 3]
b = [4, 5, 6]

a.extend(b)

print(a)
```

```python
numbers = [10, 20, 30,]

numbers.clear()

print(numbers)
```

```python
numbers = [10, 20, 30, 40]

print(numbers.index(30))
```

```python
numbers = [10, 20, 20, 30, 20]

print(numbers.count(20))
```

```python
numbers = [40, 10, 30, 20]

numbers.sort()

print(numbers)
```

```python
numbers = [40, 10, 30, 20]

numbers.sort(reverse=True)

print(numbers)
```

```python
a = [1, 2, 3]

b=a.copy()

print(b)
```

```python
numbers = [10, 20, 30, 40, 50]

print(numbers[1:4])
print(numbers[:3])
print(numbers[2:])
print(numbers[::-1])

```
